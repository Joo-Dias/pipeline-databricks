```python
# Definindo pastas do projeto em variáveis
bronze_path   = '/Volumes/bikestore/bronze/bronze_volumes'
resource_path_volume = '/Volumes/bikestore/logistics/bikestore_resource'
```

```python
df_stocks = spark.read.csv(f'{resource_path_volume}/stocks.csv',
                            header = True,
                            inferSchema = True,
                            sep = ','
                           )

display(df_stocks)
```

```python
# Salvando em formato DELTA
df_stocks.write.mode('overwrite').format('delta').option('mergeSchema', 'true').save(f'{bronze_path}/stocks')
```

```python
%sql
-- Criando uma tabela física
CREATE TABLE bikestore.bronze.bronze_stockss
USING DELTA
AS
SELECT *
FROM delta.`/Volumes/bikestore/bronze/bronze_volumes/stocks`;
```
